### Gemini-3.1-flash-lite-preview predictions on the Banking-10 test set

````python
import json
import os
import re
from concurrent.futures import ThreadPoolExecutor, as_completed

import pandas as pd
from IPython.display import display
from openai import OpenAI
from sklearn.metrics import accuracy_score, classification_report, f1_score
from tqdm.auto import tqdm

# Banking-10 test set: run Gemini on the held-out test split.
test_path = r"..\data\processed\banking-10\test\tier_200\samples.parquet"
test_df = pd.read_parquet(test_path).copy()
test_df = test_df.sample(frac=1, random_state=42).reset_index(drop=True)  # Shuffle test set

required_columns = {"text", "true_label"}
missing_columns = required_columns - set(test_df.columns)
if missing_columns:
    raise ValueError(f"Missing required columns in test set: {sorted(missing_columns)}")

allowed_labels = sorted(test_df["true_label"].astype(str).str.lower().unique().tolist())
allowed_label_block = "\n".join(f"- {label}" for label in allowed_labels)

api_key = os.getenv("GOOGLE_API_KEY") or os.getenv("GEMINI_API_KEY")
if not api_key:
    raise ValueError("Set GOOGLE_API_KEY or GEMINI_API_KEY before running this cell.")

client = OpenAI(
    api_key=api_key,
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/",
    max_retries=20,
)

REQUESTED_GEMMA_MODEL = GEMMA_MODEL = "gemma-4-31b-it"


def get_batch_label_prompt(batch_texts: list[str]) -> str:
    prompt = (
        "Classify each banking customer support message into exactly one label from the list below.\n"
        "Use the exact label text and nothing else.\n\n"
        f"Allowed labels:\n{allowed_label_block}\n\n"
        "Return the result EXCLUSIVELY as a JSON list of objects:\n"
        "[{\"label\": \"...\"}, ...]\n\n"
        "Messages to classify:\n"
    )
    for i, text in enumerate(batch_texts):
        prompt += f"\n{i + 1}. {text}"
    return prompt


def extract_json_payload(content: str) -> str:
    raw = str(content).strip()
    raw = re.sub(r"<thought>.*?</thought>", "", raw, flags=re.DOTALL | re.IGNORECASE).strip()
    fenced_match = re.search(r"```(?:json)?\s*(.*?)\s*```", raw, flags=re.DOTALL | re.IGNORECASE)
    if fenced_match:
        raw = fenced_match.group(1).strip()

    array_match = re.search(r"\[[\s\S]*\]", raw)
    if array_match:
        return array_match.group(0)

    object_match = re.search(r"\{[\s\S]*\}", raw)
    if object_match:
        return object_match.group(0)

    return raw


def parse_model_response(content: str):
    if not content or not str(content).strip():
        raise ValueError("Empty model response")

    payload = extract_json_payload(content)
    parsed = json.loads(payload)
    if isinstance(parsed, dict) and "results" in parsed:
        parsed = parsed["results"]
    if isinstance(parsed, dict):
        parsed = [parsed]
    if not isinstance(parsed, list):
        raise ValueError(f"Unexpected response type: {type(parsed).__name__}")
    return parsed


def batch_label_gemma(batch_texts: list[str]):
    for attempt in range(1, 3):
        try:
            response = client.chat.completions.create(
                model=GEMMA_MODEL,
                messages=[{"role": "user", "content": get_batch_label_prompt(batch_texts)}],
                temperature=0.0,
                max_tokens=4096,
                reasoning_effort="high",
            )
            content = response.choices[0].message.content or ""
            return parse_model_response(content)
        except Exception as exc:
            if attempt < 2:
                continue
            print(f"Error: {exc}")
            return []


def normalize_label(label: str) -> str:
    return re.sub(r"[\s_-]+", "_", str(label).strip().lower())


print(f"LLM Batch Labeling Step - Prompting {GEMMA_MODEL}...")
print(f"Test set size: {len(test_df)}")
print("Allowed labels:")
for label in allowed_labels:
    print(f"- {label}")

batch_size = 10
max_workers = 5
texts = test_df["text"].astype(str).tolist()
smoke_n = min(5, len(texts))
if smoke_n:
    smoke_results = batch_label_gemma(texts[:smoke_n])
    if len(smoke_results) != smoke_n:
        raise RuntimeError(f"Smoke test failed: expected {smoke_n} predictions, got {len(smoke_results)}")
    print(f"Smoke test passed on {smoke_n} samples.")

batch_starts = list(range(0, len(texts), batch_size))
batch_outputs: dict[int, list[dict]] = {}

with ThreadPoolExecutor(max_workers=max_workers) as executor:
    future_to_start = {
        executor.submit(batch_label_gemma, texts[start : start + batch_size]): start
        for start in batch_starts
    }
    for future in tqdm(as_completed(future_to_start), total=len(future_to_start), desc=f"Gemini labeling ({GEMMA_MODEL})"):
        start = future_to_start[future]
        try:
            batch_outputs[start] = future.result()
        except Exception as exc:
            print(f"Batch starting at {start} failed: {exc}")
            batch_outputs[start] = []

llm_results = []
for start in sorted(batch_outputs):
    llm_results.extend(batch_outputs[start])

pred_labels = [normalize_label(item.get("label", "")) for item in llm_results]
if not pred_labels:
    raise RuntimeError("Gemini did not return any valid label predictions.")

if len(pred_labels) != len(test_df):
    print(f"Warning: received {len(pred_labels)} predictions for {len(test_df)} rows. Truncating to match.")
    limit = min(len(pred_labels), len(test_df))
    eval_df = test_df.iloc[:limit].copy()
    pred_labels = pred_labels[:limit]
else:
    eval_df = test_df.copy()

eval_df["gemma_pred_label"] = pred_labels
eval_df["true_label_norm"] = eval_df["true_label"].astype(str).str.lower()

gemma_acc = accuracy_score(eval_df["true_label_norm"], eval_df["gemma_pred_label"])
gemma_f1 = f1_score(eval_df["true_label_norm"], eval_df["gemma_pred_label"], average="macro")

print(f"\nRequested {REQUESTED_GEMMA_MODEL} run using {GEMMA_MODEL} - Macro F1: {gemma_f1:.4f} | Accuracy: {gemma_acc:.4f}")
print("\nClassification report:")
print(
    classification_report(
        eval_df["true_label_norm"],
        eval_df["gemma_pred_label"],
        labels=allowed_labels,
        target_names=allowed_labels,
        zero_division=0,
    )
)

display(eval_df[["sample_id", "text", "true_label", "gemma_pred_label"]].head(10))
````

````python
from datetime import datetime, timezone
from pathlib import Path
import json
import os

import pandas as pd

run_name = f"{GEMMA_MODEL}_b{batch_size}_banking-10_test200"
run_stamp = datetime.now(timezone.utc).strftime("%Y%m%d_%H%M%S")
output_dir = Path("..") / "results" / "extended_evaluation_llm" / run_stamp / run_name
output_dir.mkdir(parents=True, exist_ok=True)

predictions_df = eval_df.drop(columns=["true_label_norm"], errors="ignore").rename(
    columns={"gemma_pred_label": "pred_label"}
).copy()

for column_name in ["confidence", "reason", "probs", "in_tokens", "out_tokens", "raw"]:
    if column_name not in predictions_df.columns:
        predictions_df[column_name] = pd.NA

raw_values = []
for item in llm_results[: len(predictions_df)]:
    try:
        raw_values.append(json.dumps(item, ensure_ascii=False))
    except TypeError:
        raw_values.append(str(item))
if len(raw_values) < len(predictions_df):
    raw_values.extend([""] * (len(predictions_df) - len(raw_values)))
predictions_df["raw"] = raw_values
predictions_df["correct"] = predictions_df["true_label"].astype(str).str.lower().eq(predictions_df["pred_label"])

predictions_columns = [
    "sample_id",
    "dataset_name",
    "text",
    "true_label",
    "meta_json",
    "pred_label",
    "confidence",
    "reason",
    "probs",
    "in_tokens",
    "out_tokens",
    "raw",
    "correct",
]
for column_name in predictions_columns:
    if column_name not in predictions_df.columns:
        predictions_df[column_name] = pd.NA
predictions_df = predictions_df[predictions_columns]
predictions_df.to_csv(output_dir / "predictions.csv", index=False)

resolved_config = {
    "name": run_name,
    "seed": int(os.getenv("SEED", "42")),
    "train_data": "data/processed/banking-10/train_seed/tier_10/samples.parquet",
    "test_data": "data/processed/banking-10/test/tier_200/samples.parquet",
    "output_dir": str(output_dir.resolve()),
    "model": {
        "kind": "google_openai_chat",
        "params": {
            "model_id": GEMMA_MODEL,
            "requested_model_id": REQUESTED_GEMMA_MODEL,
            "prompt_id": "banking-10_batch_v1",
            "batch_size": batch_size,
            "max_concurrency": max_workers,
            "temperature": 0.0,
            "max_tokens": 4096,
            "retries": 20,
            "reasoning_effort": "high",
        },
    },
}
(output_dir / "config.resolved.json").write_text(json.dumps(resolved_config, indent=2), encoding="utf-8")

report = {
    "dataset_name": "banking-10",
    "split_name": "test",
    "tier_size": int(len(predictions_df)),
    "predictor_name": f"google_openai_batch:{GEMMA_MODEL}",
    "metrics": {
        "f1_macro": float(gemma_f1),
        "accuracy": float(gemma_acc),
    },
    "extras": {
        "requested_model_id": REQUESTED_GEMMA_MODEL,
        "used_model_id": GEMMA_MODEL,
        "num_predictions": int(len(predictions_df)),
    },
    "created_at": datetime.now(timezone.utc).isoformat(),
}
(output_dir / "report.json").write_text(json.dumps(report, indent=2), encoding="utf-8")

print(f"Saved config, predictions, and report to {output_dir}")
print(f" - {output_dir / 'config.resolved.json'}")
print(f" - {output_dir / 'predictions.csv'}")
print(f" - {output_dir / 'report.json'}")
````
